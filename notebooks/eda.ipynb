```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
# Dataset directory
DATA_DIR = Path('..\data')
IMG_DIR = DATA_DIR / 'images'
MASK_DIR = DATA_DIR / 'annotations' / 'trimaps'
```

```python
# Examine images
images = list(IMG_DIR.glob('*.jpg'))
print(f'Total images: {len(images)}')
```

```python
# Examine masks
masks = [item for item in MASK_DIR.glob('*.png')
         if item.name[:2] != '._']
print(f'Total masks: {len(masks)}')
```

```python
# Visualization first 3 images and masks
fig, axes = plt.subplots(3, 2, figsize=(7, 9))

for i in range(3):
    image_path = images[i]
    image = Image.open(image_path)

    mask_path = masks[i]
    mask = Image.open(mask_path)

    axes[i, 0].imshow(image)
    axes[i, 0].set_title(f'Image: {image_path.stem}')
    axes[i, 0].axis('off')

    axes[i, 1].imshow(mask)
    axes[i, 1].set_title(f'Mask: {mask_path.stem}')
    axes[i, 1].axis('off')
plt.tight_layout()
plt.show()
```
